# With Outliers 

```python
from sklearn.model_selection import StratifiedKFold , GridSearchCV
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression()
kfold = StratifiedKFold(n_splits=5, shuffle=True, random_state=26)
param = {
    'solver':['saga','newton-cg','sag','newton-cholesky','lbfgs'],
    'C':np.linspace(0.001,5,10)
}

gcv = GridSearchCV(lr, param_grid=param, scoring='neg_log_loss', cv=kfold, n_jobs=-1)
gcv.fit(X,y)
```

```python
gcv.best_params_,gcv.best_score_
```
