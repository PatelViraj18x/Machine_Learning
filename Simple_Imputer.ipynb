```python
df = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Desktop\Machine_Learning\Cases\Chemical_Process_Data\ChemicalProcess.csv")
df.head()
```

```python
df.isna().sum().sum()
```

```python
from sklearn.impute import SimpleImputer
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import root_mean_squared_error,r2_score
```

```python
X = df.drop('Yield',axis=1)
y = df['Yield']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=26)
X_train.isnull().sum().sum(),X_test.isnull().sum().sum()
```

```python
imputer = SimpleImputer(strategy='mean').set_output(transform='pandas')
X_trn_imp = imputer.fit_transform(X_train)
X_tst_imp = imputer.transform(X_test)
X_trn_imp.isnull().sum().sum(),X_tst_imp.isnull().sum().sum()
```

```python
model = LinearRegression()
model.fit(X_trn_imp,y_train)
```

```python
y_pred = model.predict(X_tst_imp)
print('R2 Score ->',r2_score(y_test,y_pred))
print('RMSE ->',root_mean_squared_error(y_test,y_pred))
```

```python
# Ridge
import numpy as np
from sklearn.linear_model import Ridge

alpha = np.linspace(0.001,10,20)
scores = []
for x in alpha:
    ridge = Ridge(alpha=x)
    ridge.fit(X_trn_imp,y_train)
    y_pred = ridge.predict(X_tst_imp)
    scores.append([x,r2_score(y_test,y_pred),root_mean_squared_error(y_test,y_pred)])
ridge_score = pd.DataFrame(scores,columns=['Alpha','R2 Score','RMSE']).sort_values('R2 Score',ascending=False)
ridge_score.head(2)
```

```python
# Lasso
from sklearn.linear_model import Lasso

alpha = np.linspace(0.001,10,20)
scores = []
for x in alpha:
    lasso = Lasso(alpha=x)
    lasso.fit(X_trn_imp,y_train)
    y_pred = lasso.predict(X_tst_imp)
    scores.append([x,r2_score(y_test,y_pred),root_mean_squared_error(y_test,y_pred)])
lasso_score = pd.DataFrame(scores,columns=['Alpha','R2 Score','RMSE']).sort_values('R2 Score',ascending=False)
```

```python
lasso_score.head(2)
```

```python
# Elastic Net
from sklearn.linear_model import ElasticNet

alpha = np.linspace(0.001,10,20)
l1 = np.linspace(0.001,1,20)
scores = []

for x in alpha:
    for l in l1:
        elastic_net = ElasticNet(alpha=x,l1_ratio=l)
        elastic_net.fit(X_trn_imp,y_train)
        y_pred = elastic_net.predict(X_tst_imp)
        scores.append([x,l,r2_score(y_test,y_pred),root_mean_squared_error(y_test,y_pred)])
elastic_net_score = pd.DataFrame(scores,columns=['Alpha','L1_Ratio','R2 Score','RMSE']).sort_values('R2 Score',ascending=False)
```

```python
elastic_net_score.head(2)
```
