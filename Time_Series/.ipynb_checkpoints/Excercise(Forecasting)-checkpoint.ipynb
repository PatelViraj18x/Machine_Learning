```python
import pandas as pd 
import numpy as np 
import matplotlib.pyplot as plt
from sklearn.metrics import root_mean_squared_error
from tqdm import tqdm

from statsmodels.tsa.api import ExponentialSmoothing, SimpleExpSmoothing , Holt
import warnings
warnings.filterwarnings('ignore')
```

```python
df = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Desktop\Machine_Learning\Datasets\daily-total-female-births-in-cal.csv")
df.head()
```

```python
y = df['Daily total female births in California']
y_train, y_test = y.iloc[:-31], y.iloc[-31:]
y_train.shape, y_test.shape    
```

```python
from tqdm import tqdm
alpha , beta = np.linspace(0.01,0.99,10), np.linspace(0.01, 0.99, 10)
gamma = np.linspace(0.01, 0.99, 10)
scores = []
for a in tqdm(alpha):
    for b in beta:
        for g in gamma:
            for s in ['add','mul']:
                hw_add = ExponentialSmoothing(y_train, initialization_method='estimated',seasonal_periods=7, trend='add', seasonal=s)
                fit1 = hw_add.fit(smoothing_level = a, smoothing_trend=b, smoothing_seasonal=g)
                fcast1 = fit1.forecast(len(y_test))
                scores.append([a,b,g,s,root_mean_squared_error(y_test, fcast1)])
df_score = pd.DataFrame(scores, columns=['Alpha','Beta','Gamma','Seasonality','Score'])
df_score.sort_values('Score')
```

```python
alpha = 0.2
ses = SimpleExpSmoothing(y_train,initialization_method='estimated')
fit1 = ses.fit(smoothing_level = alpha)
fcast1 = fit1.forecast(len(y_test))

y_train.plot(color='blue', label='Train')
y_test.plot(color='pink', label='Test')
fcast1.plot(color='purple', label='Forecast')
plt.text(0,60, f'RMSE = {np.round(root_mean_squared_error(y_test,fcast1),2)}')
plt.legend(loc='best')
plt.show()
```

```python
alpha, beta = 0.1, 0.2
holt = Holt(y_train, initialization_method='estimated')
fit1 = holt.fit(smoothing_level = alpha, smoothing_trend=beta)
fcast1 = fit1.forecast(len(y_test))

y_train.plot(color='blue', label='Train')
y_test.plot(color='pink', label='Test')
fcast1.plot(color='purple', label='Forecast')
plt.text(0,60, f'RMSE = {np.round(root_mean_squared_error(y_test,fcast1),2)}')
plt.legend(loc='best')
plt.title('Holt\'s Linear Trend Method')
plt.show()
```

```python
alpha, beta = 0.1, 0.2
holt = Holt(y_train, initialization_method='estimated', exponential=True)
fit1 = holt.fit(smoothing_level = alpha, smoothing_trend=beta)
fcast1 = fit1.forecast(len(y_test))

y_train.plot(color='blue', label='Train')
y_test.plot(color='pink', label='Test')
fcast1.plot(color='purple', label='Forecast')
plt.text(0,60, f'RMSE = {np.round(root_mean_squared_error(y_test,fcast1),2)}')
plt.legend(loc='best')
plt.title('Holt\'s Linear Trend Method')
plt.show()
```

```python
alpha, beta, phi = 0.4, 0.2, 0.3
holt = Holt(y_train, initialization_method='estimated',damped_trend=True)
fit1 = holt.fit(smoothing_level = alpha, smoothing_trend=beta, damping_trend=phi, optimized=False)
fcast1 = fit1.forecast(len(y_test))

y_train.plot(color='blue', label='Train')
y_test.plot(color='pink', label='Test')
fcast1.plot(color='purple', label='Forecast')
plt.text(0,60, f'RMSE = {np.round(root_mean_squared_error(y_test,fcast1),2)}')
plt.legend(loc='best')
plt.title('Holt\'s Dumped Trend Method')
plt.show()
```
