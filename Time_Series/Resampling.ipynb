```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Desktop\Machine_Learning\Datasets\monthly-milk-production-pounds-p.csv",index_col=0)
df.index = pd.to_datetime(df.index).to_period('M')
df.head()
```

```python
df.index.dtype
```

```python
df.plot()
plt.title('Monthly Milk Production')
plt.xlabel('Time Series (Month)')
plt.ylabel('Production in Liters')
plt.show()
```

```python
downsampled = df.resample('Q').sum()
downsampled.index.rename('Quaeter',inplace=True)
downsampled.head()
```

```python
downsampled.plot()
plt.title('Quarterly Milk Production')
plt.xlabel('Time Series (Quarter)')
plt.ylabel('Production in Liters')
plt.show()
```

```python
df = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Desktop\Machine_Learning\Datasets\Electric_Production.csv",index_col=0)
df.index = pd.to_datetime(df.index).to_period('M')
df.head()
```

```python
df.plot()
plt.title('Monthly Electricity Production')
plt.xlabel('Time Series (Month)')
plt.ylabel('Production')
plt.show()
```

```python
downsample = df.resample('Q').sum()
downsample.index.rename('Quarter',inplace=True)
downsample.head()
```

```python
downsample.plot()
plt.title('Quarterly Electricity Production')
plt.xlabel('Time Series (Quarter)')
plt.ylabel('Production')
plt.show()
```
