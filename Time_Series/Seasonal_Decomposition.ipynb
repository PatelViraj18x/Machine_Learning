```python
from statsmodels.tsa.seasonal import seasonal_decompose

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\PGCP-AI.STUDENTSDC\Desktop\Machine_Learning\Datasets\monthly-milk-production-pounds-p.csv",index_col=0)
df.head()
```

```python
df.plot()
plt.show()
```

### Additive Decomposition

```python
series = df['Milk']
result = seasonal_decompose(series, model='additive', period=12)
result.plot()
plt.show()
```

### Multiplicative Decomposition

```python
series = df['Milk']
result = seasonal_decompose(series, model='multiplicative', period=12)
result.plot()
plt.show()
```
